## Step 4 : Handling Missing Value

```python
from sklearn.impute import SimpleImputer

df_simple = df.copy()
num_imputer = SimpleImputer(strategy = "median")
df_simple[numerical_cols] = num_imputer.fit_transform(
    df_simple[numerical_cols]
)

print("Missing values after Simple Imputer:")
print(df_simple.isnull().sum())
```

## Step 5 : Outliers

## 1.IQR method

```python
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(f"{col}: {len(outliers)} outliers using IQR")
```

## 2.Percentile Method

```python
df_percentile = df.copy()

for col in numerical_cols:
    lower = df_percentile[col].quantile(0.05)
    upper = df_percentile[col].quantile(0.95)

    outlier_count = (
        (df_percentile[col] < lower) |
        (df_percentile[col] > upper)
    ).sum()

    print(f"{col}: {outlier_count} values outside 5th-95th percentile")
```

## 3.Winsorization

```python
df_winsor = df.copy()

for col in numerical_cols:
    lower = df_winsor[col].quantile(0.05)
    upper = df_winsor[col].quantile(0.95)

    df_winsor[col] = df_winsor[col].clip(
        lower=lower,
        upper=upper
    )

print(df_winsor.head())
```

## Step 9 : Features

## Create Interaction Feature

```python
if {"total_purchases", "days_since_signup"}.issubset(df.columns):
    df["purchase_per_day"] = (
        df["total_purchases"] / df["days_since_signup"].replace(0, np.nan)
    )

print(df.head())
```

```python
from sklearn.preprocessing import FunctionTransformer

log_transformer = FunctionTransformer(np.log1p)

df_log = df.copy()

for col in numerical_cols:
    if (df_log[col].dropna() >= 0).all():
        df_log[col] = log_transformer.transform(df_log[[col]])

print(df_log.head())
```

```python
df_sqrt = df.copy()

for col in numerical_cols:
    if (df_sqrt[col].dropna() >= 0).all():
        df_sqrt[col] = np.sqrt(df_sqrt[col])

print(df_sqrt.head())
```

```python
df_reciprocal = df.copy()

for col in numerical_cols:
    df_reciprocal[col] = 1 / df_reciprocal[col].replace(0, np.nan)

print(df_reciprocal.head())
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import PowerTransformer


numerical_cols = df.select_dtypes(include="number").columns.tolist()

df_power = df.copy()

df_power[numerical_cols] = df_power[numerical_cols].replace(
    [np.inf, -np.inf], np.nan
)

df_power[numerical_cols] = df_power[numerical_cols].fillna(
    df_power[numerical_cols].median()
)

cols_to_transform = [
    col for col in numerical_cols
    if df_power[col].nunique() > 1
]
power_transformer = PowerTransformer(method="yeo-johnson")

df_power[cols_to_transform] = power_transformer.fit_transform(
    df_power[cols_to_transform]
)

print(df_power.head())
```

```python
df["amount_group"] = pd.cut(
    df["amount"],
    bins=3,
    labels=["Low", "Medium", "High"]
)

print(df[["amount", "amount_group"]].head())
```

```python
df["amount_quantile"] = pd.qcut(
    df["amount"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"],
    duplicates="drop"
)

print(df[["amount", "amount_quantile"]].head())
```

```python
threshold = df["amount"].median()

df["high_amount"] = (
    df["amount"] > threshold
).astype(int)

print(df[["amount", "high_amount"]].head())
```
